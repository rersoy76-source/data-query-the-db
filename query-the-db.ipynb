# Query the Database

Şimdi bir `ecommerce` database üzerinde gelişmiş SQL sorguları yazma zamanı!

## Data
Bu URL'de bulunan `ecommerce.sqlite` database ile çalışacağız:  
`https://wagon-public-datasets.s3.amazonaws.com/sql_databases/ecommerce.sqlite`

Dosyayı indirmek için aşağıdaki hücreyi çalıştırın.


```python
import pandas as pd
from sqlite3 import connect
```

```python
import urllib.request
import os

os.makedirs("data", exist_ok=True)

url = "https://wagon-public-datasets.s3.amazonaws.com/sql_databases/ecommerce.sqlite"

urllib.request.urlretrieve(url, "data/ecommerce.sqlite")

print("Database indirildi.")
```

```python
query_orders = """
SELECT *
FROM Orders
"""
```

## Setup

Pandas ve sqlite3 ihtiyacımız olan her şey :-)


```python
query_orders_range = """
SELECT *
FROM Orders
WHERE OrderDate BETWEEN ? AND ?
ORDER BY OrderDate
"""
```

```python
start_date = "2013-01-01"
end_date = "2013-01-31"

with connect("data/ecommerce.sqlite") as conn:
    df_orders_range = pd.read_sql(
        query_orders_range,
        conn,
        params=[start_date, end_date]
    )

df_orders_range
```

## Orders

👉 Tüm orders kayıtlarını, tüm column'ları göstererek alın.

```python
query_waiting_time = """
SELECT
    OrderID,
    julianday(ShippedDate) - julianday(OrderDate) AS waiting_time
FROM Orders
ORDER BY waiting_time
"""
```

```python
with connect("data/ecommerce.sqlite") as conn:
    df_waiting_time = pd.read_sql(
        query_waiting_time,
        conn
    )

df_waiting_time.head()
```

```python
from nbresult import ChallengeResult

result = ChallengeResult(
    "query-the-db",
    query_orders=query_orders,
    query_orders_range=query_orders_range,
    query_waiting_time=query_waiting_time
)

result.write()
print(result.check())
```

## Waiting time

👉 Delivery time değerine göre (en küçük timedelta'dan en büyüğüne doğru) artan sırada tüm orders kayıtlarını alın.

İpucu: "sqlite julianday" şeklinde arama yapın.

```python
import pandas as pd
```

Sonuç beklediğiniz gibi göründüğünde, sorgunuzu denemek ve test etmek için aşağıdaki hücreleri çalıştırın.

```python
query_orders_range = """
SELECT *
FROM Orders
WHERE OrderDate BETWEEN ? AND ?
ORDER BY OrderDate
"""
```

```python
query_waiting_time = """
SELECT
    *,
    julianday(ShippedDate) - julianday(OrderDate) AS waiting_time
FROM Orders
ORDER BY waiting_time
"""
```

```python
from nbresult import ChallengeResult

result = ChallengeResult(
    'get_waiting_time',
    query=query_waiting_time
)

result.write()
print(result.check())
```
